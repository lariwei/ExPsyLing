variance and standard deviation 

Variance: how much does each data point deviate from the mean -> we
square this because we don’t want negative values
- But! We get values from the variance that cannot be interpreted in the context
of the data
example
    the mean is 5. one point is 3 and the other point is 8
      - 5- 3= 2. so the variance for the value 3 is 2
      - 8- 5 = 2 3 so the variance is 3 
- Standard deviation: variance scaled back down to the level of the dat
