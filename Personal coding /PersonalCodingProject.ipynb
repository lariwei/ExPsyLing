```python
#this code opens the pdf and then saves it as a text file. 
#encoding overwrites the empty document tiobe.txt and puts the new readable text in
#the for loop reads each page where page means the current page. \n inserts a new line 
with pp.open("tiobe.pdf") as pdf, open("tiobe.txt" , "w", encoding = "utf-8") as f:
    for page in pdf.pages:
        t = page.extract_text()
        if t:
            f.write(t + "\n")
```
